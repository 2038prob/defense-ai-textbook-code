## 실습: PTQ 양자화와 지식 증류

### 🔬 실습 — 훈련 후 동적 양자화(PTQ)와 크기·속도 비교

① **기준 모델 준비**: 학습된 모델을 준비하고 기준선을 측정한다.


````python
import torch, torch.nn as nn, os, time, copy
model = nn.Sequential(nn.Linear(128, 256), nn.ReLU(),
                      nn.Linear(256, 64), nn.ReLU(), nn.Linear(64, 10))
model.eval()
torch.save(model.state_dict(), "fp32.pt")
print("FP32 크기(KB):", round(os.path.getsize("fp32.pt")/1024, 1))
````

② **동적 양자화**: Linear 계층을 INT8로 변환한다(보정 불필요).

````python
from torchao.quantization import (quantize_,
                                  Int8DynamicActivationInt8WeightConfig)
q_model = copy.deepcopy(model)                # 원본(FP32)은 비교용으로 남긴다
quantize_(q_model, Int8DynamicActivationInt8WeightConfig())  # 제자리 변환
torch.save(q_model.state_dict(), "int8.pt")
print("INT8 크기(KB):", round(os.path.getsize("int8.pt")/1024, 1))
````

③ **추론 속도 비교**: 동일 입력으로 평균 추론 시간을 측정한다.


````python
def bench(m, n=200):
    x = torch.randn(1, 128)
    t0 = time.time()
    with torch.no_grad():
        for _ in range(n): m(x)
    return (time.time() - t0) / n * 1000     # ms/추론
print(f"FP32 {bench(model):.3f} ms | INT8 {bench(q_model):.3f} ms")
````

**실행 결과 예시**

```
FP32 크기(KB): 198.4
INT8 크기(KB): 56.1
FP32 0.011 ms | INT8 0.200 ms
```

(크기는 코드 그대로의 실행 결과로, 약 1/4 수준이다. 정확히는 3.5분의 1인데, 스케일 같은 양자화 정보가 더해져 4분의 1보다 조금 덜 줄기 때문이다. 시간은 Apple M5 CPU(torch 2.15, torchao 0.18)에서 잰 값으로, 이 작은 모델에서는 INT8이 오히려 약 18배 느리다. 활성화를 입력마다 양자화하는 부대비용이 연산 절감분보다 크기 때문이다. 메모리는 확실히 줄지만 속도는 실제 타깃 장비에서 재 봐야 한다.)

### ONNX 내보내기와 수치 일치 검증

11.4절에서 말한 수치 일치 검증을 위 실습의 FP32 `model`로 해 본다(본문 11.6절 끝의 코드). 모델을 ONNX로 내보내고, ONNX Runtime으로 같은 입력을 추론해 PyTorch 출력과의 최대 오차를 출력한다.

````python
import numpy as np, onnxruntime as ort
x = torch.randn(1, 128)
torch.onnx.export(model, (x,), "mlp.onnx", dynamo=True,
                  input_names=["x"], output_names=["logits"],
                  external_data=False)              # 가중치까지 파일 하나로
sess = ort.InferenceSession("mlp.onnx",
                            providers=["CPUExecutionProvider"])
y_ort = sess.run(None, {"x": x.numpy()})[0]
with torch.no_grad():
    y_pt = model(x).numpy()
print("최대 오차:", float(np.abs(y_pt - y_ort).max()))
````

**실행 결과 예시**

```
최대 오차: 4.470348358154297e-08
```

(torch 2.15, onnxruntime 1.32에서 실행한 결과이다. 입력이 무작위라 값은 실행마다 조금 다르지만 부동소수점 반올림 수준인 \(10^{-7}\) 아래로 나온다. `external_data=False`로 가중치까지 `mlp.onnx` 한 파일에 담긴다.)

### 🔬 실습 — 지식 증류 학습 루프

① **KD 손실함수 정의**: 정답 CE와 교사 모방 KL(온도 \(T\))을 합해 정의한다.


````python
import torch.nn.functional as F
def kd_loss(student_logits, teacher_logits, y, T=4.0, alpha=0.5):
    ce = F.cross_entropy(student_logits, y)
    kl = F.kl_div(F.log_softmax(student_logits / T, dim=1),
                  F.softmax(teacher_logits / T, dim=1),
                  reduction="batchmean") * (T * T)     # T^2 보상
    return (1 - alpha) * ce + alpha * kl
````

② **학습 스텝**: 교사는 고정(eval), 학생만 학습한다.


````python
# --- (실습용) 교사/학생 모델과 합성 분류 데이터 준비 ---
# setup 셀의 torch, 셀 3의 nn을 사용합니다. 교사(큰 모델)/학생(작은 모델), 10-클래스 분류.
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
teacher = nn.Sequential(nn.Linear(128, 256), nn.ReLU(),
                        nn.Linear(256, 128), nn.ReLU(), nn.Linear(128, 10))
student = nn.Sequential(nn.Linear(128, 64), nn.ReLU(), nn.Linear(64, 10))
_N = 256
_X_kd = torch.randn(_N, 128)
_y_kd = torch.randint(0, 10, (_N,))
train_loader = DataLoader(TensorDataset(_X_kd, _y_kd), batch_size=64, shuffle=True)
````

````python
teacher.eval()                                # 교사 고정
optimizer = torch.optim.Adam(student.parameters(), lr=1e-3)
for x, y in train_loader:
    with torch.no_grad():
        t_logits = teacher(x)
    s_logits = student(x)
    loss = kd_loss(s_logits, t_logits, y, T=4.0, alpha=0.5)
    optimizer.zero_grad(); loss.backward(); optimizer.step()
````

**실행 결과 예시**

```
교사 정확도 0.94 (파라미터 11.2M)
학생 정확도 0.91 (파라미터 1.3M)  <- 단독 학습할 때 0.86
```

학생이 교사에 근접하며 크기는 훨씬 작다. (수치는 개념 예시이다 — 위 합성 데이터 구성에서는 교사·학생 구성에 따라 달라진다.)
